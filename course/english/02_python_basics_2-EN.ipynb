## Conditions

In a program’s flow, it is often necessary to branch depending on certain states (for example, the current value of a variable). This is done using *conditions*.

Before such a "branch," the program checks whether a condition is true or false, and then takes one path or the other.

![Condition](https://drive.google.com/uc?id=1bdvEvkfNEb2QdoWe3cCfLteAPcGWoLJ5)

Let’s imagine we are programming an ATM:

~~~
# This example uses pseudocode

amount_to_withdraw = input('How much would you like to withdraw?')

IF account_balance - amount_to_withdraw > overdraft_limit:
    dispense money
ELSE
    error message: Your account balance is insufficient
~~~

The general form of a condition in Python (and most high-level programming languages) looks like this:

~~~
if CONDITION == True:
    do this
else:  # CONDITION was not True
    do something else
~~~

The `else` part can be omitted.

### if

As an example, let’s now determine all the names from our list that are longer than 8 characters:


```python
# Note: the len() function returns the number of characters in a name
for name in names:
    if len(name) > 8:
        print(name)
```

### if ... else

With `else`, we can handle all cases that do not meet the condition specified in `if`. In the following `else` section, we want to count how many names are shorter than or equal to 8 characters:

```python
# we define two variables in which we store the number of
# short and long names, respectively
number_short_names = 0
number_long_names = 0

# we iterate through all the names in the list and increase the number stored
# in the respective variable by 1 whenever we find a name that meets
# the corresponding condition.
for name in names:
    if len(name) > 8:
        number_long_names = number_long_names + 1
    else:
        number_short_names = number_short_names + 1

# at the end, we print the number of short and long names
print(f"{number_short_names} short names and {number_long_names} long names")
```

### Subconditions

If conditions can be nested:

```python
number_short_names = 0
number_medium_names = 0
number_long_names = 0

for name in names:
    # Name is longer than 8 characters
    if len(name) > 8:
        number_long_names = number_long_names + 1
    # Name is shorter than 9 characters
    else:
        if len(name) < 5:
            number_short_names = number_short_names + 1
        # here we count names with length >= 5 (because of the second "if")
        # and length <= 8 (because of the first "if")
        else:
            number_medium_names = number_medium_names + 1
print(f'{number_short_names} short names, {number_medium_names} names of medium length and {number_long_names} long names')
```

### if ... elif ... else

In Python, such nested conditions can often be avoided by using `elif`. Python goes through the sequence of conditions until the first one evaluates to `True`. All subsequent `elif`s and the `else` are then ignored:

```python
number_short_names = 0
number_medium_names = 0
number_long_names = 0

for name in names:
    if len(name) > 8:
        number_long_names = number_long_names + 1
    elif len(name) < 5:
        number_short_names = number_short_names + 1
    else:
        number_medium_names = number_medium_names + 1

print(f'{number_short_names} short names, {number_medium_names} names of medium length {number_long_names} long names')
```

### The `in` Operator
Some names appear multiple times in the list `names`. Depending on the question, this may or may not be desired. Let’s try to prevent duplicate names. For this, we need to introduce a new operator that tests whether a value is present in a sequence: `in`.

```python
'a' in 'Anakonda'
```

`in` works with all iterables. Since lists are iterables, the `in` operator also works with lists. Here, we check whether the integer `42` appears in a list:

```python
42 in [1, 55, 44, 32, 71, 41]
```

In the next example, we use the `in` operator to check whether the name already appears in a list of unique names:

```python
unique_names = []

for name in names:
    if name in unique_names:
        pass  # do nothing
    else:
        unique_names.append(name)
print(f'List names: {len(names)} entries, list unique_names: {len(unique_names)} entries')
```

The `pass` in the fourth line of this example is a special feature of Python. After a colon (`if name in unique_names:`), there must be at least one statement. In the specific case of our example, nothing needs to be done if the name is already in `unique_names`. However, because of the colon in the previous line, we need to put a statement. This is exactly why Python has the `pass` statement. It is the equivalent of an empty pair of curly brackets in other programming languages:

~~~
if(condition) {
}
~~~

## Functions

Functions are “subprograms” that are called from the main program. The advantages of using functions are:

* Functions structure the code and make it easier to understand.
* Functions are reusable: A function written once can be called multiple times at different places in a program. In principle, a function can also be “recycled” in other programs.
* Functions prevent redundancy: Each functionality appears only once in the program. This makes the program easier to maintain and reduces the likelihood of errors.

We recognize a function by its name, followed by parentheses. Typically, the function is *applied to an object*. The name of the object is given in the parentheses. This is also called the *argument* of the function.

```python
# here we apply the function "len()" to the object "list"
a_list = [1, 2, 3, 4]
len(a_list)
```

During the course, we have already encountered some functions:

* `print()` – outputs a string
* `type()` – returns the type of an object
* `len()` – returns the length of an object

In each case, a function is called that someone else kindly wrote for us and that is available in every Python installation. However, we can also define our own functions.

### Writing a Function

A function definition in Python begins with the keyword `def` (many other languages use `function` instead). After `def` comes the name of the function. Ideally, this should be a verb, since a function always performs an action. The function name is followed by a pair of parentheses and a colon. After that, indented, comes the actual *function body*, which defines what the function does.

```python
def say_hello():
    print('Hello!')
```

With this, we have written a function named `say_hello`. The function has been created during execution—our notebook now knows it exists—but it does not do anything yet because we have not called it. Calling the function looks like this:

```python
say_hello()
```

### Function Parameters

The function `say_hello()` that we just wrote represents only the minimal version of a function: it always does the same thing. We can make the function more flexible by assigning **parameters** to it during its definition:

```python
def say_hello(name):
    print(f'Hello {name}!')
```

Here, the function definition specifies that a *parameter* must be passed to the function when it is called. This value is then available within the function as the variable `name`. We can pass this value as an *argument* when calling the function:

```python
say_hello('Otto')
```

```python
say_hello('Anna')
```

### Return Values

Every function returns a value when called. This returned value essentially replaces the function call.

The return value is specified using the `return` statement and should always appear at the end of the *function body*.

```python
# define a function that takes a number as a parameter and
# returns the square of the number
def square(number):
    return number * number
```

```python
square(2) + square(3)  # 4 + 9
```

If no `return` value is explicitly specified, the return value is always `None`, i.e., the "empty" value. Here is an example that illustrates this implicitly returned value:

```python
def say_hello(name):
    print(f'Hello {name}!')
```

```python
return_value = say_hello('Otto')  # saves the return value in a variable
print(f'Return value: {return_value}')
```

### Explicit Return Values

Explicit return values are always useful when a function, for example, performs a calculation and the result of that calculation needs to be used in the main program.

The following function transforms the given string so that only the first and last character are returned. Between these two characters, the number of omitted characters is placed. For example, `abcd` becomes `a2d`.

```python
def shorten(long_string):
    # only useful for strings with more than 2 characters
    if len(long_string) > 2:

        # return the first (long_string[0]) and last character
        # (long_string[-1]) and in between the number
        # (len(long_string) - 2) of omitted characters
        return_value = f"{long_string[0]}{len(long_string) - 2}{long_string[-1]}"

    # if the string is short anyway, we simply return the string itself
    else:
        return_value = long_string
    return return_value
```

```python
shorten("Christopher")
```

### Functions with Multiple Parameters

In principle, a function can have any number of parameters. In practice, unless there are very good reasons, one should limit it to a maximum of 3 or 4 parameters, as this makes the function easier to use (i.e., easier to understand).

In the following example, we calculate the volume of a cuboid:

```python
def calculate_volume(length, width, height):
  """Calculates the volume of a cuboid.

     Arguments
     length:   length in m
     width:    width in m
     height:   height in m

     Returns
     vol:      volume in m³
  """

  vol = length * width * height
  return vol
```

```python
calculate_volume(0.8, 0.4, 0.4)
```

### Docstrings

In the function definition above, we saw a so-called "docstring" for the first time. It is usually used to describe the functionality of functions in detail: what the function does, which data types it expects as input parameters, and what it returns.

A docstring is a string that can span multiple lines. It is enclosed with triple quotes <tt>"""</tt>.

We can view the docstrings of functions using the help function:

```python
?calculate_volume
```

### Default Parameters

Normally, the order of the arguments when calling a function must match the order of the parameters in the function declaration, and all arguments must be provided:

~~~
def set_userdata(username, age):
    ...
    
set_userdata('otto', 20)    
~~~

If not all parameters are provided when calling the function, this will result in an error:

```python
def set_userdata(name, age):
    return name, age
```

```python
set_userdata('Hans')
```

If we do not require a parameter to be provided, we can assign a default value to it:

```python
def set_userdata(name, age=None):
    return name, age
```

**Note**: In most programming languages, functions can return only a single value. In Python, however (as in the function above), two or more values can be returned directly.

```python
print(set_userdata('otto', 20))
print(set_userdata('anna'))
```

If the value is provided, it is used in the function; otherwise, the default value is used. Such preassigned parameters must appear **after** all other parameters, which is why the following code does not work.

```python
def set_userdata(age=None, name):
    return name, age
```

If more than one default parameter is defined, individual arguments can be addressed by their name when calling the function ("keyword arguments"), so we no longer have to follow the given order:

```python
def set_userdata(name, age=None, weight=None, height=None):
    return name, age, weight, height
```

```python
print(set_userdata('Otto', height=181))
print(set_userdata('Otto', height=181, age=25))
```

This is a pattern we will see very often: in machine learning, we frequently use functions that are already predefined and have many parameters. Very often, we only change a few of them and leave the other parameters at their default values, which are usually already set sensibly.
