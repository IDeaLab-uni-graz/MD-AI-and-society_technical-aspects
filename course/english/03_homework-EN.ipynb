```python
# import relevant modules
from sklearn.neighbors import KNeighborsClassifier
import seaborn as sns

# filter and prepare data
titanic_data = sns.load_dataset("titanic")
titanic_data = titanic_data.dropna(subset=["age", "fare", "survived"]).copy()
features = ["age", "fare"]
label = "survived"
X = titanic_data[features]
y = titanic_data[label]

# train classifier
knn_classifier = KNeighborsClassifier()
knn_classifier.fit(X, y)

# make predictions and store in table
predictions = knn_classifier.predict(X)
titanic_data["prediction"] = predictions

# visualize predictions
sns.scatterplot(data=titanic_data, x="age", y="fare", hue="prediction")
```
