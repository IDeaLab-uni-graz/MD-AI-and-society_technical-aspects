## Naive Bayes

In the [lecture](https://janalasser.at/lectures/MD_KI/VO1_3_algorithms_supervised_learning/#/1/0/17), we learned about frequency-based guessing (also called the "Naive Bayes" algorithm) as a useful baseline for classification algorithms. Below, we implement the example from the lecture predicting wine quality in Python:

```python
# import Pandas as "pd" to handle the data
import pandas as pd

# we create a table with data on wine quality and features
colors = ['red','white','white','red','white']
age = ['old', 'old', 'young', 'old', 'young']
quality = ['good', 'good', 'good', 'bad', 'bad']
wine_data = pd.DataFrame({'color': colors, 'age': age, 'quality': quality})

# finally, we display the table
wine_data
```

With this data, we still cannot work, since the dataset contains strings instead of numbers. As already explained in the [lecture on Data Transformations](https://janalasser.at/lectures/MD_KI/VO3_3_data_transformations/#/1), we first need to convert these categorical variables into numbers. This is done in `pandas` using the `map()` function.

We pass a dictionary to the `map()` function (see also the documentation on dictionaries in Python in [Chapter 5.5](https://docs.python.org/3/tutorial/datastructures.html)), which maps each categorical value to a number.

```python
# the value "red" is mapped to the number 1, the value "white" to 0
wine_data['color'] = wine_data['color'].map({'red': 1, 'white': 0})

# the value "old" is mapped to the number 1, the value "young" to 0
wine_data['age'] = wine_data['age'].map({'old': 1, 'young': 0})

# the value "good" is mapped to the number 1, the value "bad" to 0
wine_data['quality'] = wine_data['quality'].map({"good": 1, "bad": 0})

# this is how the table, i.e., the DataFrame, looks now:
wine_data
```

Each observation (i.e., each wine) is represented by a **row** in the table. Now we need to clarify which are the **features** and which are the **labels** (or classes, or "target values") of the observations. Following the [nomenclature from the lecture](https://janalasser.at/lectures/MD_KI/VO1_2_supervised_learning/#/1/0/2), we assign the features to a (multidimensional!) variable named `x` and the labels to a variable named `y`.

```python
features = ["color", "age"]
label = "quality"

x = wine_data[features]
y = wine_data[label]
```

```python
# this is how the variable "x" with the observations looks
x
```

```python
# and this is how the variable "y" with the labels or target values looks
y
```

We now want to train a classification algorithm with the existing observations (see [classification pipeline](https://janalasser.at/lectures/MD_KI/VO1_2_supervised_learning/#/2/0/2)) to predict the quality of the wine for a new observation.

For this, we use the `CategoricalNB` function provided by `scikit-learn`. This stands for "Categorical Naive Bayes" (see also Chapter 1.9.5 of the [documentation](https://scikit-learn.org/1.5/modules/naive_bayes.html#gaussian-naive-bayes) of `scikit-learn` for the mathematical background, or the [function documentation](https://scikit-learn.org/1.5/modules/generated/sklearn.naive_bayes.CategoricalNB.html#sklearn.naive_bayes.CategoricalNB) itself).

`scikit-learn` offers several versions of the Naive Bayes algorithm for data following different distributions. Since all our features are categorical variables, the `Categorical` version is suitable for our use case.

```python
# first, we import the corresponding module that provides the
# CategoricalNB() function
import sklearn.naive_bayes

# then we create an "empty" classifier by calling the
# function. At this point, we could also provide parameters
# that modify the behavior of the classifier. However, the
# default parameter values are sufficient for our use case
naive_bayes_classifier = sklearn.naive_bayes.CategoricalNB()

# finally, we train the classifier with the
# observation data. This is also called "fitting".
naive_bayes_classifier.fit(x, y)
```

We now want to apply the trained classifier to predict the quality of the wine for a new observation, for which we only know the feature values (i.e., color and age). For this, we first create another table with the new observation.

**Note**: Although we are creating a table with only a single row here, there is nothing stopping us from storing multiple new observations in the table and classifying them all at once.

```python
# for the new observation, we only know the features "color" and "age" -
# not the label, i.e., the quality.
colors = ['red']
age = ['young']

# we create a DataFrame again
new_observation = pd.DataFrame({'color': colors, 'age': age})
# and convert the categorical data into numerical values
new_observation['color'] = new_observation['color'].map({'red': 1, 'white': 0})
new_observation['age'] = new_observation['age'].map({'old': 1, 'young': 0})

# this is how the table with the new observation looks now
new_observation
```

To predict the quality of the new wine, we use the `predict()` function provided by the trained classifier, which takes the new observation as an argument and outputs the prediction of the target value.

```python
naive_bayes_classifier.predict(new_observation)
```

## K-Nearest Neighbors

Another classification algorithm that we learned about in the [lecture](https://janalasser.at/lectures/MD_KI/VO1_3_algorithms_supervised_learning/#/2/0/9) is the K-Nearest Neighbors algorithm. It uses the classes of surrounding neighbors to predict the class of a data point.

Below, we implement K-Nearest Neighbors classification with the Titanic dataset from the [previous session](https://colab.research.google.com/drive/1fnJAvk72jWjAAsJWqxX6fznH91Jdai3E?usp=sharing). Here, we are interested in predicting which passengers survived.

```python
# as the first step, we import the dataset as before
import seaborn as sns
titanic_data = sns.load_dataset("titanic")
titanic_data.head()
```

```python
# for now, we are only interested in three columns of the dataset:
# whether a passenger survived, the age, and the ticket fare
interesting_columns = ["age", "fare", "survived"]

# we filter out the interesting columns
filtered_titanic_data = titanic_data[interesting_columns]
filtered_titanic_data
```

```python
# we notice that this dataset still contains "NaN" values. This causes
# a problem for the K-Nearest Neighbors algorithm. So first, we remove
# all observations (passengers) for which any value in the table is "NaN"
# using the dropna() function and make a copy of the dataset
# with the copy() function.
filtered_titanic_data = filtered_titanic_data.dropna().copy()
filtered_titanic_data
```

```python
# reminder: this is what the distribution of age and ticket fare
# looks like in the dataset
import seaborn as sns
sns.scatterplot(data=filtered_titanic_data, x="age", y="fare")
```

```python
# we again specify which columns contain the features and which
# column is the label or target value that we want to predict
features = ["age", "fare"]
label = "survived"

x = filtered_titanic_data[features]
y = filtered_titanic_data[label]
```

```python
# again, scikit-learn provides a pre-built function that implements the
# K-Nearest Neighbors algorithm, which we can use after importing
# the corresponding module
from sklearn.neighbors import KNeighborsClassifier
knn_classifier = KNeighborsClassifier(n_neighbors=35)
```

```python
# we train the classifier with the observations
knn_classifier.fit(x, y)

# and then predict (for the existing observations!) whether the
# respective passengers survived
predictions = knn_classifier.predict(x)
```

```python
# to easily compare the predictions with the actual survival
# information of the passengers, we add them to the table
filtered_titanic_data["prediction"] = predictions
filtered_titanic_data.head()
```

```python
# we can also visualize for which observations our algorithm predicts
# "survived" (orange) or "did not survive" (blue) by using the "hue"
# argument of the scatterplot() function.
sns.scatterplot(data=filtered_titanic_data, x="age", y="fare", hue="prediction")
```
