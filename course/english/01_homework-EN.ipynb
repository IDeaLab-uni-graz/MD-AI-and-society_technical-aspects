## Loops

### Exercise Loops 1

Try slowly increasing the value of the variable `loops_per_loop` in the following program — for example, from 10 to 100, 200, 300, and so on. Remember: as long as a `*` appears next to the cell, Python is busy running the loops. Stop the process if it takes too long. This exercise is meant to show how quickly a large number of loop iterations can occur when loops are nested.

**Note**: `counter += 1` is the same as `counter = counter + 1`. In other words, both expressions mean "increase the variable counter by 1". This notation is used to shorten the code.

```python
loops_per_loop = 10 # number of iterations per loop

counter = 0
for i in range(loops_per_loop):
    for j in range(loops_per_loop):
        for k in range(loops_per_loop):
            counter += 1
print(f"{counter} iterations")
```

### Exercise Loops 2

In the course, we learned about the `for` loop, which iterates over all elements of a sequence (or, more generally, an iterable). In addition, there is the `while` loop, which checks a condition in each iteration and stops when that condition is no longer met.

In the following example, the condition is that the current number is not 10. As soon as the compared number is 10, the `while` loop ends:

```python
i = 0
while i != 10:
    print(i)
    i += 1
```

### Exercise Lists 2 – Multidimensional Lists

We have seen that a list can contain values of any data type, including other lists. This allows us to create a list whose elements are themselves lists — a list of lists.

Suppose we measure the temperature three times a day and want to store these measurements. On the first day, we have these three measurements: `[17, 28, 24]`. On the second day, we measure `[18, 31, 28]`. On the third day, we measure `[20, 35, 29]`. So, for each day, we have a list with three values. We can then store these daily lists inside another list:

```python
temperatures = [
    [17, 28, 24],
    [18, 31, 28],
    [20, 35, 29]
]
```

We can think of these temperatures as a table: each row represents a day, and each column represents a measurement time (e.g., 6:00, 12:00, 18:00). We have already learned how to access the measurements for a specific day:

```python
temperatures[1]
```

Since the element accessed this way is itself a list, we can also access individual elements within it. For example, we can read the first measurement of the second day like this:

```python
day = temperatures[1]       # temperatures of day two as list
first_temperature = day[0]  # first value of the day
first_temperature
```

Normally, this is written more compactly like this:

```python
temperatures[1][0]
```

The value in the first set of brackets refers to the list `temperatures` (i.e., what we referred to above as `day`), and the value in the second set of brackets refers to the list obtained via the first index — that is, the first entry of the second element of the list `temperatures`.

To get familiar with this concept, print the following temperatures:

* the second temperature of the first day
* the first temperature of the first day
* the first temperature of the last day
* the last temperature of the second day

**Note**: Such multidimensional lists are very similar to the tables we learned about in the [lecture](https://janalasser.at/lectures/MD_KI/VO2_4_data/), where we store data for machine learning.
