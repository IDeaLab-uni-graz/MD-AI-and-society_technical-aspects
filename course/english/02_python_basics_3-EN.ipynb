## Modules

### What Are Modules?

Python organizes code into *modules*.
A module is nothing more than a file with the `.py` extension.

Modules are used to:

* Organize large projects into smaller, more manageable files. Logically related pieces of code (e.g., functions and variables) are stored together in a single file (a *module*).
* Make code more reusable, since modules can be selectively *imported* into new code.

Small scripts can easily fit into a single file.
However, once a program has more than a few hundred lines, it is recommended to split the code into multiple modules, also because this makes the code easier to maintain and test.

### Modules Must Be Imported

To use a module in your own program, the module must first be imported:

~~~
import <module_name>
~~~

After that, the module and its defined functionality are available.

### The Standard Library

Not only can your own program code be organized into modules, but external code is also distributed in modules. Every Python installation comes with the so-called "standard library." It contains a set of modules that are always available to us.

The module `random`, included in the standard library, provides a number of functions for randomization. To use them, we first need to import the module:

```python
import random
```

After that, we can, for example, use the function <tt>randint()</tt> to generate a random integer:


```python
random.randint(0, 100)
```

### Modules and Namespaces

In the last example, we didn’t use just any function named `randint`, but specifically the one provided by the `random` module.

Modules therefore not only structure code, but also create *namespaces*, which prevent, for example, two functions with the same name defined in different modules from overwriting each other.

The namespace is simply formed by the module name. In the example above, that is <tt>random</tt>.

### Renaming Namespaces
Some namespace names are very long, making it tedious to type them repeatedly. Therefore, it is possible to assign a custom name to a module. This is done when we import the module. We assign the new name using the word <tt>as</tt>:


```python
import random as rnd
```

```python
rnd.randint(0, 100)
```

## Handling Data

The third-party library <tt>pandas</tt> is our main tool for working with datasets. It helps to load and transform datasets and is also processed by default by functions for machine learning.

### DataFrames

In the code below, we manually create a table—a so-called <tt>DataFrame</tt> with two columns: names and age. Normally, we will read DataFrames from a file or use datasets provided as examples by <tt>pandas</tt> or other libraries.

```python
# import Pandas under the abbreviation "pd"
import pandas as pd

# we create a table for testing that contains names
# (strings) and ages (numbers)

# first we create the list "names" and
# the list "ages"
names = ['tim','olaf','nina','andrea','jana','miro']
ages = [31, 52, 14, 46, 28, 3]

# then we create the table "people"
# (a DataFrame) and pass the two
# lists as columns of the table
people = pd.DataFrame({'name':names, 'age':ages})

# finally we display the first six entries
# of the table
people.head(6)
```

**Note**: to create the <tt>DataFrame</tt>, we used a data structure that we haven’t learned yet: the "dictionary." This data structure is denoted by curly brackets <tt>{ }</tt> and consists of pairs of "keys" and "values" that are each connected by a colon:

<tt>
{  
'name':names,  
'age':ages  
}
</tt>

You can find more about this in the [documentation on dictionaries](https://docs.python.org/3/tutorial/datastructures.html).

### Descriptive Statistics

We can access individual columns of the <tt>DataFrame</tt> by using the column names:

```python
people['name']
```

```python
people['age']
```

We can also display which columns are present in the <tt>DataFrame</tt>:

```python
people.columns
```

We can also get the number of rows using the <tt>len()</tt> function.

```python
len(people)
```

<tt>DataFrames</tt> already provide a lot of practical functionality for working with data. For example, we can easily calculate the mean of a column:

```python
people['age'].mean()
```

We can also easily obtain a comprehensive summary of the information contained in a column:

```python
people['age'].describe()
```

### Filtering DataFrames

One method of filtering <tt>DataFrames</tt> we have already learned is selecting specific columns.

Often, however, we are more interested in selecting certain *rows*. We can achieve this by excluding rows using a *mask*. Let’s recall the <tt>DataFrame</tt> with the names and ages of people from earlier:

```python
people.head(6)
```

We can now use *Boolean values* (see the [notebook](https://colab.research.google.com/drive/1pwvNv70fzHxR2t1hzmSc1zU0chNAJak1?usp=sharing) on Python basics from last time, chapter "The Data Type bool") to include (<tt>True</tt>) or exclude (<tt>False</tt>) individual rows. To construct the mask, we create a list of Boolean values:

```python
# here we construct the mask
mask = [False, True, False, True, False, True]

# we can now use the mask to select only certain
# rows in the DataFrame:
people[mask]
```

### Complex Filters

We can now use this knowledge to construct a filter that checks two conditions simultaneously:

```python
mask = (people['age'] > 30) & (people['age'] < 50)
mask
```

**Note**: pay attention to the parentheses around the logical expressions. These are important to ensure the correct order of evaluation: first the expressions inside the parentheses are evaluated from left to right, and then the operations between them.

```python
middle_aged_people = people[mask]
middle_aged_people.head()
```

Of course, we can also combine different columns in such filters. Here is an example with the Titanic dataset, where we filter out all male passengers in second class:

```python
mask = (data_titanic['sex'] == 'male') & (data_titanic['class'] == 'Second')
men_second_class = data_titanic[mask]
men_second_class.head()
```

## Visualizing Data

Visualizing data helps us understand what information is contained in a dataset. Depending on whether we are dealing with numerical or categorical data (see also [Lecture on Data Types](https://janalasser.at/lectures/MD_KI/VO2_4_data/)), we need different types of charts to represent the data.

For the rest of the course, we will need four types of charts:

* **Histogram**: represents a one-dimensional dataset with numerical data points.
* **Bar plot**: represents a two-dimensional dataset with one numerical and one categorical dimension.
* **Scatter plot**: represents a two-dimensional dataset with two numerical dimensions.
* **Line plot**: represents a two-dimensional dataset with two numerical dimensions and is used when we want to emphasize changes in one of the dimensions. A classic example is the change of a variable over time.

To visualize data in this course, we will use the third-party library <tt>[seaborn](https://seaborn.pydata.org/index.html)</tt>. Seaborn provides a relatively high-level interface with functions for creating charts and is built on top of the powerful <tt>[matplotlib](https://matplotlib.org/)</tt> library. For most use cases, <tt>seaborn</tt> is sufficient – but if you want to tweak the details of your plots, it can be helpful to work with <tt>matplotlib</tt> and use its functionalities.

```python
# import the seaborn library for creating
# plots from data
import seaborn as sns
```

### Histogram

All functions in <tt>seaborn</tt> follow the same pattern:

* The function name determines the type of plot (e.g., histogram or bar plot).
* The first parameter of the function is the dataset to be visualized.
* Depending on whether the plot is one- or two-dimensional, the `x` and `y` parameters specify which column of the dataset should be used for each dimension.
* Additional optional parameters can be used to adjust the behavior or appearance of the plot.

```python
# since histograms only represent one-dimensional data,
# we only need to provide the column name for the x-dimension
# to the histplot() function
sns.histplot(data_titanic, x="age")
```

### Line Plot

To meaningfully demonstrate a line plot, we need data with a time dimension. The COVID infection numbers from earlier are suitable.

We read the data again. It is important to ensure that the column containing the date is also interpreted (parsed) as a date (`parse_dates=['date']`).

```python
data_url = "https://github.com/owid/covid-19-data/raw/master/public/data/jhu/full_data.csv"
data_COVID = pd.read_csv(data_url, parse_dates=['date'])

# we are only interested in the data from Austria, so we first filter
# the dataset with the case numbers accordingly
mask = data_COVID['location'] == 'Austria'
data_AUT = data_COVID[mask]
```

For the line plot, we again need to specify two dimensions, where the y-dimension represents the variable whose changes we are interested in. The x-dimension is the variable along which the changes occur – in this example, time.

```python
sns.lineplot(data=data_AUT, x="date", y="new_cases")
```
