## Create, edit and execute cells

The notebook consists of individual **cells**, which can contain either text or code (more on this later).


### Modal Editor

The Jupyter Notebook has a **modal** user interface. This means that keyboard inputs trigger different actions depending on which **mode** the notebook is currently in. There are two modes:
1. Edit mode
2. Command mode

In **edit mode**, a row of text formatting options appears above the cell in Google Colab. In **command mode**, the formatting options disappear.

<font color='green'>Activate **edit mode** by</font>
<ul class="outside">  
<li><font color='green'>pressing <b>enter</b> while the mouse pointer is on the cell, or</font></li>  
<li><font color='green'>double-clicking inside a cell with the mouse, or</font></li>  
<li><font color='green'>clicking <b>Edit</b> in the cell’s menu.</font></li>  
</ul>

<font color='green'>Activate **command mode** by</font>
<ul class="outside">  
<li><font color='green'>pressing <b>esc</b> while the mouse pointer is on the cell, or</font></li>  
<li><font color='green'>clicking with the mouse to the left outside the cell.</font></li>  
</ul>

When the cell is in edit mode, you can modify its content just like in a regular text editing program, such as Microsoft Word.

When the cell is in command mode, you can edit the notebook as a whole—for example, you can delete, move, copy, and paste entire cells—but you cannot modify the content of individual cells. To do that, use the menu that appears above the cell on the right once you select a cell.
